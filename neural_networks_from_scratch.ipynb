```python
import mlcroissant as mlc
import pandas as pd

# Fetch the Croissant JSON-LD
croissant_dataset = mlc.Dataset('https://www.kaggle.com/datasets/uciml/red-wine-quality-cortez-et-al-2009/croissant/download')

# Check what record sets are in the dataset
record_sets = croissant_dataset.metadata.record_sets
print(record_sets)

# Fetch the records and put them in a DataFrame
red_wine = pd.DataFrame(croissant_dataset.records(record_set=record_sets[0].uuid))
red_wine.head()
```

```python
red_wine.columns = red_wine.columns.str.replace('winequality-red.csv/', '', regex=False).str.replace('+', '_', regex=False)
display(red_wine.head())
```

```python
red_wine.shape # (rows, columns)
```

```python
# Create training and validation splits
df_train = red_wine.sample(frac=0.7, random_state=0)
df_valid = red_wine.drop(df_train.index)
display(df_train.head(4))

# Scale to [0, 1]
max_ = df_train.max(axis=0)
min_ = df_train.min(axis=0)
df_train = (df_train - min_) / (max_ - min_)
df_valid = (df_valid - min_) / (max_ - min_)

# Split features and target
X_train = df_train.drop('quality', axis=1)
X_valid = df_valid.drop('quality', axis=1)
y_train = df_train['quality']
y_valid = df_valid['quality']
```

**Capa Densamente Conectada (Dense layer):**

Capa donde cada neurona está conectada a todas las neuronas de la capa anterior.

Cómo funciona:
- Cada conexión tiene un peso que se ajusta durante el entrenamiento
- Combina las entradas multiplicándolas por los pesos y sumando los resultados
- Añade un "sesgo" (bias) para ajustar la salida

Ejemplo de aplicación:
- Reconocimiento de imágenes (como capas intermedias)
- Predicción de precios de viviendas
- Sistemas de recomendación

Limitaciones:
- Puede ser computacionalmente costoso con muchas neuronas
- Riesgo de sobreajuste (overfitting) si no se regulariza
- No captura relaciones espaciales como las capas convolucionales

**Función de Activación ReLU (Rectified Linear Unit):**

Una función que convierte todos los números negativos a cero y deja los positivos igual.

Limitaciones:

- Problema de "neuronas muertas" (si siempre da 0, no aprende)
- No es adecuada para capas de salida en muchos casos
- No es diferenciable en x=0 (aunque en práctica no suele ser problema)

```python
from tensorflow import keras
from tensorflow.keras import layers, callbacks

# Detiene el entrenamiento cuando la métrica de validación deja de mejorar.
early_stopping = callbacks.EarlyStopping(
    min_delta=0.001, # Sólo considera mejoras mayores a 0.001
    patience=20, # Espera 20 épocas sin mejora antes de detener
    restore_best_weights=True, # Restaura los pesos del modelo al mejor estado encontrado
)
# Sequential: Indica que es un modelo de capas apiladas linealmente (entrada → capas ocultas → salida).
# Capas Dense: Son capas totalmente conectadas donde cada neurona está conectada a todas las neuronas de la capa anterior.
model = keras.Sequential([
    layers.Dense(512, activation='relu', input_shape=[11]),
    layers.Dense(512, activation='relu'),
    layers.Dense(512, activation='relu'),
    layers.Dense(1),
])
model.compile(
    optimizer='adam',
    loss='mae',
)
```

```python
history = model.fit(
    X_train, y_train,
    # El rendimiento del modelo se evalúa con estos datos al final de cada época,
    # pero el modelo no aprende de ellos. Esto ayuda a detectar el sobreajuste.
    validation_data=(X_valid, y_valid),
    batch_size=256, # Procesa 256 muestras antes de actualizar pesos
    epochs=500, # Número de veces que el modelo iterará sobre todo el conjunto de datos de entrenamiento.
    callbacks=[early_stopping],
    verbose=0,  # Esta configuración suprime la salida del progreso del entrenamiento en la consola.
)

history_df = pd.DataFrame(history.history)
history_df.loc[:, ['loss', 'val_loss']].plot();
# Graficas de pérdidas en entrenamiento (loss) y la pérdida de validación (val_loss) a lo largo de las épocas.
# Underfitting: Ambas curvas altas y no convergen
#cOverfitting: Brecha creciente entre train y validación
# Buen ajuste: Ambas convergen a valores bajos
print("Minimum validation loss: {}".format(history_df['val_loss'].min()))
```

# **Funciones de activacion**

**Relu:**
- La función más utilizada en capas ocultas
- Computacionalmente eficiente (simple operación de umbral)
- No tiene salidas negativas
- Puede tener problema de neuronas muertas, es decir, dejar de activarse (siempre salen 0)

**Elu:**
- Redes donde se observa el problema de neuronas muertas.
- Cuando se necesitan salidas negativas
- Redes más pequeñas donde el costo computacional adicional no es problema

**SELU:**
- Redes profundas totalmente conectadas
- Arquitecturas auto-normalizantes
- Cuando se quiere evitar técnicas como Batch Normalization, técnica que normaliza las salidas de una capa (media ≈ 0, varianza ≈ 1) antes de aplicar la función de activación.

**Sigmoid:**
- Capas de salida en clasificación binaria
- Redes pequeñas

**Tanh:**
- Capas ocultas en RNNs
- Cuando se necesitan salidas en (-1, 1)

**Swish:**
- Redes donde ReLU no proporciona buen desempeño
- Modelos donde se puede permitir el costo computacional adicional
- En pruebas experimentales para mejorar métricas

El gradiente es un vector que nos indica la dirección en la que deben ir los pesos. Más precisamente, nos indica cómo modificar los pesos para que la pérdida cambie más rápido. Llamamos a nuestro proceso descenso de gradiente porque utiliza el gradiente para descender la curva de pérdida hacia un mínimo. Estocástico significa "determinado por el azar". Nuestro entrenamiento es estocástico porque los minilotes son muestras aleatorias del conjunto de datos. ¡Por eso se llama SGD!

La muestra de datos de entrenamiento de cada iteración se denomina minibatch "minilote" (o, a menudo, simplemente batch "lote"), mientras que una ronda completa de datos de entrenamiento se denomina epoch "época". El número de épocas de entrenamiento determina cuántas veces la red verá cada ejemplo de entrenamiento.

**Tasa de aprendizaje y tamaño del lote**

La magnitud de estos desplazamientos está determinada por la tasa de aprendizaje. Una tasa de aprendizaje menor significa que la red neuronal necesita ver más lotes antes de que sus pesos convergan a sus valores óptimos.

La tasa de aprendizaje y el tamaño de los lotes son los dos parámetros que mayor efecto tienen en el desarrollo del entrenamiento Gradiente Descendiente Estocastico.

Para la mayoría de los trabajos no será necesario realizar una búsqueda exhaustiva de hiperparámetros para obtener resultados satisfactorios. Adam es un algoritmo SGD con una tasa de aprendizaje adaptativa que lo hace adecuado para la mayoría de los problemas sin necesidad de ajustar los parámetros (es autoajustable, en cierto sentido). Adam es un excelente optimizador de propósito general.

```python
## Animacion de gradiente descendente estocastico
import matplotlib.pyplot as plt
import numpy as np
from IPython import display

def animate_sgd(learning_rate, batch_size, num_examples, steps, true_w, true_b):
    # Generar datos sinteticos
    X = np.random.rand(num_examples, 1)
    noise = np.random.randn(num_examples, 1) * 0.1
    y = true_w * X + true_b + noise
    # Inicializar pesos y sesgo
    w = np.random.randn(1, 1)
    b = np.random.randn(1, 1)
    # Graficar
    fig, ax = plt.subplots()
    ax.scatter(X, y, label='Data')
    line, = ax.plot(X, w * X + b, color='red', label='Model')
    ax.set_xlabel('X')
    ax.set_ylabel('y')
    ax.set_title('Stochastic Gradient Descent Animation')
    ax.legend()
    # SGD loop
    for step in range(steps):
        # Seleccionar un lote aleatorio
        indices = np.random.choice(num_examples, size=batch_size)
        X_batch = X[indices]
        y_batch = y[indices]
        # Calcular gradientes
        dw = (2/batch_size) * np.sum(X_batch * (w * X_batch + b - y_batch))
        db = (2/batch_size) * np.sum(w * X_batch + b - y_batch)
        # Actualizar pesos y sesgos
        w -= learning_rate * dw
        b -= learning_rate * db
        # Actualizar grafico
        line.set_ydata(w * X + b)
        ax.set_title(f'Step {step+1}, Loss: {np.mean((w * X + b - y)**2):.4f}')
        display.clear_output(wait=True)
        display.display(fig)
        plt.show()
    plt.close(fig)
```

```python
learning_rate = 0.03
batch_size = 128
num_examples = 512

animate_sgd(
    learning_rate=learning_rate,
    batch_size=batch_size,
    num_examples=num_examples,
    # You can also change these, if you like
    steps=20, # total training steps (batches seen)
    true_w=3.0, # the slope of the data
    true_b=2.0, # the bias of the data
)
```

```python
import nnfs
from nnfs.datasets import spiral_data
nnfs.init()
## Capa densa
class Layer_Dense:
  ## Capa de inicializacion
  def __init__(self, n_inputs, n_neurons):
    ## Inicializar pesos y sesgos
    self.weights = 0.01 * np.random.randn(n_inputs, n_neurons)
    self.biases = np.zeros((1, n_neurons))

  ## Paso hacia adelante
  def forward(self, inputs):
    ## Calcular valores de salida de acuerdo a la entrada, pesos y sesgos
    self.output = np.dot(inputs, self.weights) + self.biases

## Crear espiral
X, y = spiral_data(samples = 100, classes = 3)
## Crear capa densa con 2 caracteristicas de entrad y 3 valores de salida
dense1 = Layer_Dense(2, 3)
## Desarrollar pasos hacia adelante con datos de entrenamiento a traves de esta capa
dense1.forward(X)

## Salida de la primera capa
print(dense1.output[:5])
```
